3. Playfair Cipher

```python
def create_matrix(key):
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
    chars = ""

    for ch in (key + alphabet).upper():
        if ch == 'J':
            ch = 'I'
        if ch in alphabet and ch not in chars:
            chars += ch

    return [list(chars[i:i+5]) for i in range(0, 25, 5)]


def prepare_text(text):
    text = ''.join(
        ch for ch in text.upper() if ch.isalpha()
    ).replace('J', 'I')

    pairs = []
    i = 0

    while i < len(text):
        a = text[i]

        if i + 1 == len(text):
            b = 'X'
            i += 1
        elif text[i] == text[i + 1]:
            b = 'X'
            i += 1
        else:
            b = text[i + 1]
            i += 2

        pairs.append((a, b))

    return pairs


def encrypt_pair(a, b, matrix):
    for r in range(5):
        for c in range(5):
            if matrix[r][c] == a:
                r1, c1 = r, c
            if matrix[r][c] == b:
                r2, c2 = r, c

    if r1 == r2:
        return matrix[r1][(c1+1) % 5] + matrix[r2][(c2+1) % 5]
    elif c1 == c2:
        return matrix[(r1+1) % 5][c1] + matrix[(r2+1) % 5][c2]
    else:
        return matrix[r1][c2] + matrix[r2][c1]


key = input("Enter keyword: ")
text = input("Enter plaintext: ")

matrix = create_matrix(key)
pairs = prepare_text(text)

print("5x5 Matrix:")
for row in matrix:
    print(' '.join(row))

cipher = ""
for a, b in pairs:
    cipher += encrypt_pair(a, b, matrix)

print("Ciphertext:", cipher)
```

9. Decrypt a Playfair Cipher (PT-109 Message)

```python
def decrypt_playfair(cipher, key):
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
    key = key.upper().replace("J", "I")

    matrix = ""
    for ch in key + alphabet:
        if ch in alphabet and ch not in matrix:
            matrix += ch

    cipher = "".join(
        ch for ch in cipher.upper() if ch in alphabet
    )

    result = ""

    for i in range(0, len(cipher) - 1, 2):
        a, b = cipher[i], cipher[i + 1]
        r1, c1 = divmod(matrix.index(a), 5)
        r2, c2 = divmod(matrix.index(b), 5)

        if r1 == r2:
            result += matrix[r1 * 5 + (c1 - 1) % 5]
            result += matrix[r2 * 5 + (c2 - 1) % 5]
        elif c1 == c2:
            result += matrix[((r1 - 1) % 5) * 5 + c1]
            result += matrix[((r2 - 1) % 5) * 5 + c2]
        else:
            result += matrix[r1 * 5 + c2]
            result += matrix[r2 * 5 + c1]

    return result


cipher = """KXJEY UREBE ZWEHE WRYTU HEYFS
KREHE GOYFI WTTTU OLKSY CAJPO
BOTEI ZONTX BYBNT GONEY CUZWR
GDSON SXBOU YWRHE BAAHY USEDQ"""

key = input("Enter the Playfair keyword: ")
print("Decrypted message:", decrypt_playfair(cipher, key))
```

10. Encrypt a Message Using the Given Playfair Matrix

```python
def encrypt_playfair(text):
    matrix = [
        "MFHIK",
        "UNOPQ",
        "ZVWXY",
        "ELARG",
        "DSTBC"
    ]

    text = "".join(
        ch for ch in text.upper() if ch.isalpha()
    ).replace("J", "I")

    pairs = []
    i = 0

    while i < len(text):
        a = text[i]
        if i + 1 < len(text):
            b = text[i + 1]
            if a == b:
                pairs.append(a + "X")
                i += 1
            else:
                pairs.append(a + b)
                i += 2
        else:
            pairs.append(a + "X")
            i += 1

    result = ""

    for a, b in pairs:
        r1, c1 = next(
            (r, row.index(a))
            for r, row in enumerate(matrix) if a in row
        )
        r2, c2 = next(
            (r, row.index(b))
            for r, row in enumerate(matrix) if b in row
        )

        if r1 == r2:
            result += matrix[r1][(c1 + 1) % 5]
            result += matrix[r2][(c2 + 1) % 5]
        elif c1 == c2:
            result += matrix[(r1 + 1) % 5][c1]
            result += matrix[(r2 + 1) % 5][c2]
        else:
            result += matrix[r1][c2]
            result += matrix[r2][c1]

    return result


message = "Must see you over Cadogan West. Coming at once."
print("Plaintext:", message)
print("Ciphertext:", encrypt_playfair(message))
```

12. Hill Cipher — Encryption and Decryption

```python
import numpy as np

K = np.array([[9, 4], [5, 7]])
K_inv = np.array([[3, 24], [9, 9]])

text = input("Enter plaintext: ").upper()
text = "".join(ch for ch in text if ch.isalpha())

if len(text) % 2 != 0:
    text += "X"

cipher = ""

for i in range(0, len(text), 2):
    p = np.array([
        ord(text[i]) - 65,
        ord(text[i + 1]) - 65
    ])

    c = K @ p % 26
    cipher += chr(c[0] + 65) + chr(c[1] + 65)

print("Ciphertext:", cipher)

plain = ""

for i in range(0, len(cipher), 2):
    c = np.array([
        ord(cipher[i]) - 65,
        ord(cipher[i + 1]) - 65
    ])

    p = K_inv @ c % 26
    plain += chr(p[0] + 65) + chr(p[1] + 65)

print("Decrypted text:", plain)
```

13. Known-plaintext attack on the Hill cipher

```python
import numpy as np

# Two plaintext pairs: HI and BY
P = np.array([[7, 1], [8, 24]])

# Corresponding ciphertext pairs (example)
C = np.array([[10, 5], [11, 2]])

det = int(round(np.linalg.det(P))) % 26

if det == 0:
    print("Choose different plaintext pairs.")
else:
    det_inv = next(
        (i for i in range(26) if det * i % 26 == 1),
        None
    )

    if det_inv is None:
        print("Plaintext matrix has no modular inverse.")
    else:
        adj = np.array([
            [P[1, 1], -P[0, 1]],
            [-P[1, 0], P[0, 0]]
        ])

        P_inv = (det_inv * adj) % 26
        K = (C @ P_inv) % 26

        print("Recovered key matrix:")
        print(K)
```

15. Letter frequency attack on an additive cipher

```python
from collections import Counter

cipher = input("Enter ciphertext: ").upper()
n = int(input("How many results? "))

freq = [
    8.2, 1.5, 2.8, 4.3, 12.7, 2.2, 2.0, 6.1, 7.0,
    0.15, 0.77, 4.0, 2.4, 6.7, 7.5, 1.9, 0.1, 6.0,
    6.3, 9.1, 2.8, 1.0, 2.4, 0.15, 2.0, 0.07
]

def decrypt(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            result += chr((ord(ch) - 65 - shift) % 26 + 65)
        else:
            result += ch
    return result

def score(text):
    letters = [ch for ch in text if ch.isalpha()]
    if not letters:
        return float("inf")

    counts = Counter(letters)
    total = len(letters)

    return sum(
        (counts.get(chr(65 + i), 0) / total * 100 - freq[i]) ** 2
        / freq[i]
        for i in range(26)
    )

results = []

for shift in range(26):
    plain = decrypt(cipher, shift)
    results.append((score(plain), shift, plain))

results.sort()

for s, shift, plain in results[:n]:
    print("Shift:", shift, "|", plain)
```

21. ECB, CBC and CFB Modes

```python

def encrypt(block, key):
    return block ^ key

def decrypt(block, key):
    return block ^ key

def ecb(data, key):
    return [encrypt(x, key) for x in data]

def cbc(data, key, iv):
    result = []
    for x in data:
        c = encrypt(x ^ iv, key)
        result.append(c)
        iv = c
    return result

def cfb(data, key, iv):
    result = []
    for x in data:
        c = x ^ encrypt(iv, key)
        result.append(c)
        iv = c
    return result

data = [10, 20, 30, 40]
key = 15
iv = 25

print("Plaintext:", data)

print("ECB:", ecb(data, key))
print("CBC:", cbc(data, key, iv))
print("CFB:", cfb(data, key, iv))
```

22. CBC Mode Encryption and Decryption Using S-DES

```python

# S-DES tables
P10 = [3,5,2,7,4,10,1,9,8,6]
P8 = [6,3,7,4,8,5,10,9]
IP = [2,6,3,1,4,8,5,7]
IP_INV = [4,1,3,5,7,2,8,6]
EP = [4,1,2,3,2,3,4,1]
P4 = [2,4,3,1]

S0 = [[1,0,3,2],[3,2,1,0],
      [0,2,1,3],[3,1,3,2]]
S1 = [[0,1,2,3],[2,0,1,3],
      [3,0,1,0],[2,1,0,3]]

def perm(x, table):
    return ''.join(x[i-1] for i in table)

def xor(a, b):
    return ''.join('1' if x != y else '0'
                   for x, y in zip(a, b))

def keys(k):
    p = perm(k, P10)
    l, r = p[:5], p[5:]
    l, r = l[1:]+l[:1], r[1:]+r[:1]
    k1 = perm(l+r, P8)
    l, r = l[2:]+l[:2], r[2:]+r[:2]
    k2 = perm(l+r, P8)
    return k1, k2

def fk(x, k):
    l, r = x[:4], x[4:]
    z = xor(perm(r, EP), k)

    def sb(v, s):
        row = int(v[0]+v[3], 2)
        col = int(v[1:3], 2)
        return format(s[row][col], '02b')

    out = sb(z[:4], S0) + sb(z[4:], S1)
    return xor(l, perm(out, P4)) + r

def encrypt(x, k):
    k1, k2 = keys(k)
    x = perm(x, IP)
    x = fk(x, k1)
    x = x[4:] + x[:4]
    x = fk(x, k2)
    return perm(x, IP_INV)

def decrypt(x, k):
    k1, k2 = keys(k)
    x = perm(x, IP)
    x = fk(x, k2)
    x = x[4:] + x[:4]
    x = fk(x, k1)
    return perm(x, IP_INV)

def cbc_encrypt(data, k, iv):
    result = ""
    for i in range(0, len(data), 8):
        block = data[i:i+8]
        c = encrypt(xor(block, iv), k)
        result += c
        iv = c
    return result

def cbc_decrypt(data, k, iv):
    result = ""
    for i in range(0, len(data), 8):
        block = data[i:i+8]
        p = xor(decrypt(block, k), iv)
        result += p
        iv = block
    return result

key = "0111111101"
iv = "10101010"
plain = "0000000100100011"

cipher = cbc_encrypt(plain, key, iv)
print("Encrypted:", cipher)

decoded = cbc_decrypt(cipher, key, iv)
print("Decrypted:", decoded)
```

23. Counter Mode (CTR) Encryption and Decryption Using S-DES

```python

def ctr_encrypt(data, key):
    result = ""

    for i in range(0, len(data), 8):
        block = data[i:i+8]

        counter = format(i // 8, '08b')
        stream = encrypt(counter, key)

        result += xor(block, stream)

    return result

def ctr_decrypt(data, key):
    return ctr_encrypt(data, key)

key = "0111111101"
plain = "000000010000001000000100"

cipher = ctr_encrypt(plain, key)
print("Encrypted:", cipher)

decoded = ctr_decrypt(cipher, key)
print("Decrypted:", decoded)
```

30. CBC-MAC – Forgery Attack

```python

def encrypt(block, key):
    return block ^ key

def cbc_mac(blocks, key):
    result = 0

    for block in blocks:
        result = encrypt(result ^ block, key)

    return result

key = 25

# Original one-block message
X = 10

# Calculate original MAC
T = cbc_mac([X], key)

print("Original message:", X)
print("Original MAC:", T)

# Create forged two-block message
X2 = X ^ T

forged_message = [X, X2]

# Calculate forged MAC
forged_mac = cbc_mac(forged_message, key)

print("Forged message:", forged_message)
print("Forged MAC:", forged_mac)
```

31. CMAC Subkey Generation

```python

def left_shift(data, size):
    msb = data >> (size - 1)
    result = (data << 1) & ((1 << size) - 1)

    if size == 64:
        R = 0x1B
    else:
        R = 0x87

    if msb:
        result ^= R

    return result


def cmac_subkeys(size, encrypted_zero):
    k1 = left_shift(encrypted_zero, size)
    k2 = left_shift(k1, size)

    return k1, k2


size = 128
encrypted_zero = int(
    "7DF76B0C1AB899B33E42F047B91B546F", 16
)

k1, k2 = cmac_subkeys(size, encrypted_zero)

print("K1:", format(k1, '032X'))
print("K2:", format(k2, '032X'))
```

32. DSA – Different Signatures for the Same Message

```python

import hashlib

p = 23
q = 11
g = 4
x = 3

message = "HELLO"

h = int(hashlib.sha256(
    message.encode()
).hexdigest(), 16) % q

def dsa_sign(k):
    y = pow(g, k, p) % q
    r = y
    s = (pow(k, -1, q) * (h + x*r)) % q
    return r, s

sig1 = dsa_sign(2)
sig2 = dsa_sign(3)

print("Message:", message)
print("Signature 1:", sig1)
print("Signature 2:", sig2)
print("Signatures different:", sig1 != sig2)
```

34. ECB, CBC and CFB Modes with Padding

```python

def pad(data, size):
    data = data + "1"
    while len(data) % size != 0:
        data += "0"
    return data

def ecb(data, key, size):
    result = ""
    for i in range(0, len(data), size):
        block = data[i:i+size]
        result += format(
            int(block, 2) ^ key, f'0{size}b'
        )
    return result

data = input("Enter binary plaintext: ")
size = 8
key = 15

padded = pad(data, size)

print("Original:", data)
print("Padded:", padded)
print("ECB Ciphertext:", ecb(padded, key, size))
```

35. One-Time Pad Version of Vigenère Cipher

```python

text = input("Enter plaintext: ").upper()
key = list(map(int, input("Enter key numbers: ").split()))

result = ""

if len(key) != len(text.replace(" ", "")):
    print("Key length must match the number of letters.")
elif any(k < 0 or k > 25 for k in key):
    print("Key values must be between 0 and 25.")
else:
    j = 0

    for ch in text:
        if ch.isalpha():
            p = ord(ch) - ord('A')
            c = (p + key[j]) % 26
            result += chr(c + ord('A'))
            j += 1
        else:
            result += ch

    print("Ciphertext:", result)
```

36. Affine Caesar Cipher

```python

import math

text = input("Enter plaintext: ").upper()
a = int(input("Enter a: "))
b = int(input("Enter b: "))

if math.gcd(a, 26) != 1:
    print("Invalid key! a must be coprime with 26.")
else:
    result = ""

    for ch in text:
        if ch.isalpha():
            p = ord(ch) - ord('A')
            c = (a*p + b) % 26
            result += chr(c + ord('A'))
        else:
            result += ch

    print("Ciphertext:", result)
```

37. Letter Frequency Attack on Monoalphabetic Substitution Cipher

```python

from collections import Counter

text = input("Enter ciphertext: ").upper()
top = int(input("Enter number of possible plaintexts: "))

english = "ETAOINSHRDLCUMWFGYPBVKJXQZ"

letters = [ch for ch in text if ch.isalpha()]
frequency = Counter(letters)

sorted_cipher = sorted(
    frequency, key=frequency.get, reverse=True
)

print("\nPossible plaintexts:")

for shift in range(min(top, 26)):
    mapping = {}

    for i, ch in enumerate(sorted_cipher):
        mapping[ch] = english[(i + shift) % 26]

    result = ""

    for ch in text:
        result += mapping.get(ch, ch)

    print(shift + 1, ".", result)
```

38. Hill Cipher – Known Plaintext Attack

```python

import sympy as sp

def matrix_inverse_mod(A, mod):
    det = int(A.det()) % mod

    if sp.gcd(det, mod) != 1:
        raise ValueError("Matrix is not invertible modulo 26")

    inv_det = pow(det, -1, mod)
    adj = A.adjugate()

    return (inv_det * adj).applyfunc(
        lambda x: int(x) % mod
    )

# Two known plaintext blocks: HI and BY
P = sp.Matrix([
    [7, 1],
    [8, 24]
])

# Corresponding ciphertext blocks
C = sp.Matrix([
    [19, 13],
    [22, 3]
])

try:
    P_inv = matrix_inverse_mod(P, 26)
    K = (C * P_inv).applyfunc(lambda x: int(x) % 26)

    print("Recovered Hill key:")
    print(K)
except ValueError as e:
    print(e)
```

39. Letter Frequency Attack on Additive Cipher

```python

from collections import Counter

text = input("Enter ciphertext: ").upper()
top = int(input("Enter top results: "))

english = "ETAOINSHRDLCUMWFGYPBVKJXQZ"

def decrypt(text, key):
    result = ""
    for ch in text:
        if ch.isalpha():
            p = (ord(ch) - ord('A') - key) % 26
            result += chr(p + ord('A'))
        else:
            result += ch
    return result

def score(text):
    letters = [c for c in text if c.isalpha()]
    freq = Counter(letters)

    order = sorted(freq, key=freq.get, reverse=True)

    return sum(
        abs(order.index(ch) - english.index(ch))
        for ch in order
    )

results = []

for key in range(26):
    plain = decrypt(text, key)
    results.append((score(plain), key, plain))

results.sort()

print("\nPossible plaintexts:")

for s, k, plain in results[:top]:
    print("Key:", k, "Plaintext:", plain)
```
