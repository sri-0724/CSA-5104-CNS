```python

text = input("Enter plaintext: ")
key = int(input("Enter key (1-25): "))

if key < 1 or key > 25:
    print("Invalid key. Enter a value from 1 to 25.")
else:
    cipher = ""

    for ch in text:
        if ch.isupper():
            cipher += chr((ord(ch) - ord('A') + key) % 26 + ord('A'))
        elif ch.islower():
            cipher += chr((ord(ch) - ord('a') + key) % 26 + ord('a'))
        else:
            cipher += ch

    print("Encrypted text:", cipher)
```

```python

plain_alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
cipher_alphabet = "QWERTYUIOPASDFGHJKLZXCVBNM"

text = input("Enter plaintext: ")
cipher = ""

for ch in text:
    if ch.isupper():
        index = plain_alphabet.index(ch)
        cipher += cipher_alphabet[index]
    elif ch.islower():
        index = plain_alphabet.index(ch.upper())
        cipher += cipher_alphabet[index].lower()
    else:
        cipher += ch

print("Encrypted text:", cipher)
```

```python

def create_matrix(keyword):
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
    key = ""

    for ch in keyword.upper():
        if ch.isalpha():
            ch = ch.replace("J", "I")
            if ch not in key:
                key += ch

    for ch in alphabet:
        if ch not in key:
            key += ch

    matrix = []
    for i in range(0, 25, 5):
        matrix.append(list(key[i:i + 5]))

    return matrix


def prepare_text(text):
    text = "".join(
        ch for ch in text.upper() if ch.isalpha()
    ).replace("J", "I")

    result = ""
    i = 0

    while i < len(text):
        first = text[i]

        if i + 1 >= len(text):
            second = "X"
            i += 1
        elif text[i] == text[i + 1]:
            second = "X"
            i += 1
        else:
            second = text[i + 1]
            i += 2

        result += first + second

    return result


def encrypt(text, matrix):
    result = ""

    for i in range(0, len(text), 2):
        a = text[i]
        b = text[i + 1]

        for r in range(5):
            for c in range(5):
                if matrix[r][c] == a:
                    r1, c1 = r, c
                if matrix[r][c] == b:
                    r2, c2 = r, c

        if r1 == r2:
            result += matrix[r1][(c1 + 1) % 5]
            result += matrix[r2][(c2 + 1) % 5]
        elif c1 == c2:
            result += matrix[(r1 + 1) % 5][c1]
            result += matrix[(r2 + 1) % 5][c2]
        else:
            result += matrix[r1][c2]
            result += matrix[r2][c1]

    return result


keyword = input("Enter keyword: ")
plaintext = input("Enter plaintext: ")

matrix = create_matrix(keyword)
prepared = prepare_text(plaintext)

print("5 x 5 Matrix:")
for row in matrix:
    print(" ".join(row))

print("Prepared text:", prepared)
print("Encrypted text:", encrypt(prepared, matrix))
```

```python

plaintext = input("Enter plaintext: ")
key = input("Enter key: ").upper()

if not key.isalpha():
    print("Invalid key. Use letters only.")
else:
    cipher = ""
    key_index = 0

    for ch in plaintext:
        if ch.isalpha():
            shift = ord(key[key_index % len(key)]) - ord('A')

            if ch.isupper():
                encrypted = chr(
                    (ord(ch) - ord('A') + shift) % 26 + ord('A')
                )
            else:
                encrypted = chr(
                    (ord(ch) - ord('a') + shift) % 26 + ord('a')
                )

            cipher += encrypted
            key_index += 1
        else:
            cipher += ch

    print("Encrypted text:", cipher)
```

```python

from math import gcd

plaintext = input("Enter plaintext: ")
a = int(input("Enter value of a: "))
b = int(input("Enter value of b: "))

if gcd(a, 26) != 1:
    print("Invalid value of a. Choose a coprime with 26.")
else:
    cipher = ""

    for ch in plaintext:
        if ch.isupper():
            p = ord(ch) - ord('A')
            c = (a * p + b) % 26
            cipher += chr(c + ord('A'))
        elif ch.islower():
            p = ord(ch) - ord('a')
            c = (a * p + b) % 26
            cipher += chr(c + ord('a'))
        else:
            cipher += ch

    print("Encrypted text:", cipher)
```
