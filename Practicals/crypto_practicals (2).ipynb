1. Caesar Cipher

```python
text = input("Enter plaintext: ")
k = int(input("Enter key (1-25): "))
result = ""

for ch in text:
    if ch.isupper():
        result += chr((ord(ch) - ord('A') + k) % 26 + ord('A'))
    elif ch.islower():
        result += chr((ord(ch) - ord('a') + k) % 26 + ord('a'))
    else:
        result += ch

print("Ciphertext:", result)
```

2. Monoalphabetic Substitution Cipher

```python
text = input("Enter plaintext: ")
key = input("Enter 26-letter substitution key: ").upper()

if len(key) != 26 or len(set(key)) != 26 or not key.isalpha():
    print("Invalid key!")
else:
    result = ""
    for ch in text:
        if ch.isupper():
            result += key[ord(ch) - ord('A')]
        elif ch.islower():
            result += key[ord(ch) - ord('a')].lower()
        else:
            result += ch

    print("Ciphertext:", result)
```

3. Playfair Cipher

```python
def create_matrix(key):
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
    chars = ""

    for ch in (key + alphabet).upper():
        if ch == 'J':
            ch = 'I'
        if ch in alphabet and ch not in chars:
            chars += ch

    return [list(chars[i:i+5]) for i in range(0, 25, 5)]


def prepare_text(text):
    text = ''.join(
        ch for ch in text.upper() if ch.isalpha()
    ).replace('J', 'I')

    pairs = []
    i = 0

    while i < len(text):
        a = text[i]

        if i + 1 == len(text):
            b = 'X'
            i += 1
        elif text[i] == text[i + 1]:
            b = 'X'
            i += 1
        else:
            b = text[i + 1]
            i += 2

        pairs.append((a, b))

    return pairs


def encrypt_pair(a, b, matrix):
    for r in range(5):
        for c in range(5):
            if matrix[r][c] == a:
                r1, c1 = r, c
            if matrix[r][c] == b:
                r2, c2 = r, c

    if r1 == r2:
        return matrix[r1][(c1+1) % 5] + matrix[r2][(c2+1) % 5]
    elif c1 == c2:
        return matrix[(r1+1) % 5][c1] + matrix[(r2+1) % 5][c2]
    else:
        return matrix[r1][c2] + matrix[r2][c1]


key = input("Enter keyword: ")
text = input("Enter plaintext: ")

matrix = create_matrix(key)
pairs = prepare_text(text)

print("5x5 Matrix:")
for row in matrix:
    print(' '.join(row))

cipher = ""
for a, b in pairs:
    cipher += encrypt_pair(a, b, matrix)

print("Ciphertext:", cipher)
```

4. Polyalphabetic Substitution Cipher (Vigenère)

```python
text = input("Enter plaintext: ")
key = input("Enter keyword: ").upper()

result = ""
j = 0

for ch in text:
    if ch.isalpha():
        shift = ord(key[j % len(key)]) - ord('A')

        if ch.isupper():
            result += chr((ord(ch) - ord('A') + shift) % 26 + ord('A'))
        else:
            result += chr((ord(ch) - ord('a') + shift) % 26 + ord('a'))

        j += 1
    else:
        result += ch

print("Ciphertext:", result)
```

5. Affine Caesar Cipher

```python
import math

text = input("Enter plaintext: ")
a = int(input("Enter a: "))
b = int(input("Enter b: "))

if math.gcd(a, 26) != 1:
    print("Invalid key! gcd(a, 26) must be 1.")
else:
    result = ""

    for ch in text:
        if ch.isupper():
            p = ord(ch) - ord('A')
            result += chr((a * p + b) % 26 + ord('A'))
        elif ch.islower():
            p = ord(ch) - ord('a')
            result += chr((a * p + b) % 26 + ord('a'))
        else:
            result += ch

    print("Ciphertext:", result)
```

9. Decrypt a Playfair Cipher (PT-109 Message)

```python
def decrypt_playfair(cipher, key):
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
    key = key.upper().replace("J", "I")

    matrix = ""
    for ch in key + alphabet:
        if ch in alphabet and ch not in matrix:
            matrix += ch

    cipher = "".join(
        ch for ch in cipher.upper() if ch in alphabet
    )

    result = ""

    for i in range(0, len(cipher) - 1, 2):
        a, b = cipher[i], cipher[i + 1]
        r1, c1 = divmod(matrix.index(a), 5)
        r2, c2 = divmod(matrix.index(b), 5)

        if r1 == r2:
            result += matrix[r1 * 5 + (c1 - 1) % 5]
            result += matrix[r2 * 5 + (c2 - 1) % 5]
        elif c1 == c2:
            result += matrix[((r1 - 1) % 5) * 5 + c1]
            result += matrix[((r2 - 1) % 5) * 5 + c2]
        else:
            result += matrix[r1 * 5 + c2]
            result += matrix[r2 * 5 + c1]

    return result


cipher = """KXJEY UREBE ZWEHE WRYTU HEYFS
KREHE GOYFI WTTTU OLKSY CAJPO
BOTEI ZONTX BYBNT GONEY CUZWR
GDSON SXBOU YWRHE BAAHY USEDQ"""

key = input("Enter the Playfair keyword: ")
print("Decrypted message:", decrypt_playfair(cipher, key))
```

10. Encrypt a Message Using the Given Playfair Matrix

```python
def encrypt_playfair(text):
    matrix = [
        "MFHIK",
        "UNOPQ",
        "ZVWXY",
        "ELARG",
        "DSTBC"
    ]

    text = "".join(
        ch for ch in text.upper() if ch.isalpha()
    ).replace("J", "I")

    pairs = []
    i = 0

    while i < len(text):
        a = text[i]
        if i + 1 < len(text):
            b = text[i + 1]
            if a == b:
                pairs.append(a + "X")
                i += 1
            else:
                pairs.append(a + b)
                i += 2
        else:
            pairs.append(a + "X")
            i += 1

    result = ""

    for a, b in pairs:
        r1, c1 = next(
            (r, row.index(a))
            for r, row in enumerate(matrix) if a in row
        )
        r2, c2 = next(
            (r, row.index(b))
            for r, row in enumerate(matrix) if b in row
        )

        if r1 == r2:
            result += matrix[r1][(c1 + 1) % 5]
            result += matrix[r2][(c2 + 1) % 5]
        elif c1 == c2:
            result += matrix[(r1 + 1) % 5][c1]
            result += matrix[(r2 + 1) % 5][c2]
        else:
            result += matrix[r1][c2]
            result += matrix[r2][c1]

    return result


message = "Must see you over Cadogan West. Coming at once."
print("Plaintext:", message)
print("Ciphertext:", encrypt_playfair(message))
```

12. Hill Cipher — Encryption and Decryption

```python
import numpy as np

K = np.array([[9, 4], [5, 7]])
K_inv = np.array([[3, 24], [9, 9]])

text = input("Enter plaintext: ").upper()
text = "".join(ch for ch in text if ch.isalpha())

if len(text) % 2 != 0:
    text += "X"

cipher = ""

for i in range(0, len(text), 2):
    p = np.array([
        ord(text[i]) - 65,
        ord(text[i + 1]) - 65
    ])

    c = K @ p % 26
    cipher += chr(c[0] + 65) + chr(c[1] + 65)

print("Ciphertext:", cipher)

plain = ""

for i in range(0, len(cipher), 2):
    c = np.array([
        ord(cipher[i]) - 65,
        ord(cipher[i + 1]) - 65
    ])

    p = K_inv @ c % 26
    plain += chr(p[0] + 65) + chr(p[1] + 65)

print("Decrypted text:", plain)
```

13. Known-plaintext attack on the Hill cipher

```python
import numpy as np

# Two plaintext pairs: HI and BY
P = np.array([[7, 1], [8, 24]])

# Corresponding ciphertext pairs (example)
C = np.array([[10, 5], [11, 2]])

det = int(round(np.linalg.det(P))) % 26

if det == 0:
    print("Choose different plaintext pairs.")
else:
    det_inv = next(
        (i for i in range(26) if det * i % 26 == 1),
        None
    )

    if det_inv is None:
        print("Plaintext matrix has no modular inverse.")
    else:
        adj = np.array([
            [P[1, 1], -P[0, 1]],
            [-P[1, 0], P[0, 0]]
        ])

        P_inv = (det_inv * adj) % 26
        K = (C @ P_inv) % 26

        print("Recovered key matrix:")
        print(K)
```

15. Letter frequency attack on an additive cipher

```python
from collections import Counter

cipher = input("Enter ciphertext: ").upper()
n = int(input("How many results? "))

freq = [
    8.2, 1.5, 2.8, 4.3, 12.7, 2.2, 2.0, 6.1, 7.0,
    0.15, 0.77, 4.0, 2.4, 6.7, 7.5, 1.9, 0.1, 6.0,
    6.3, 9.1, 2.8, 1.0, 2.4, 0.15, 2.0, 0.07
]

def decrypt(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            result += chr((ord(ch) - 65 - shift) % 26 + 65)
        else:
            result += ch
    return result

def score(text):
    letters = [ch for ch in text if ch.isalpha()]
    if not letters:
        return float("inf")

    counts = Counter(letters)
    total = len(letters)

    return sum(
        (counts.get(chr(65 + i), 0) / total * 100 - freq[i]) ** 2
        / freq[i]
        for i in range(26)
    )

results = []

for shift in range(26):
    plain = decrypt(cipher, shift)
    results.append((score(plain), shift, plain))

results.sort()

for s, shift, plain in results[:n]:
    print("Shift:", shift, "|", plain)
```

16. Frequency attack on a monoalphabetic substitution cipher

```python
from collections import Counter

cipher = input("Enter ciphertext: ").upper()
n = int(input("How many results? "))

english = "ETAOINSHRDLCUMWFGYPBVKJXQZ"

letters = [ch for ch in cipher if ch.isalpha()]
counts = Counter(letters)

sorted_cipher = sorted(
    counts, key=counts.get, reverse=True
)

mapping = {}

for i, ch in enumerate(sorted_cipher):
    mapping[ch] = english[i]

plain = ""

for ch in cipher:
    if ch.isalpha():
        plain += mapping.get(ch, ch)
    else:
        plain += ch

print("Frequency-based candidate:")
print(plain)
print("Requested number of candidates:", n)
```

18. DES key generation using two 28-bit halves

```python
key56 = input("Enter 56-bit key: ")

if len(key56) != 56 or any(b not in "01" for b in key56):
    print("Enter exactly 56 binary bits.")
else:
    C = key56[:28]
    D = key56[28:]

    shifts = [1, 1, 2, 2, 2, 2, 2, 2,
              1, 2, 2, 2, 2, 2, 2, 1]

    for i, shift in enumerate(shifts, 1):
        C = C[shift:] + C[:shift]
        D = D[shift:] + D[:shift]

        # Each half is 28 bits.
        # PC-2 selects 24 bits from C and 24 from D.
        print("Round", i)
        print("C:", C)
        print("D:", D)
```

19. Encryption using CBC mode with 3DES

```python
from Crypto.Cipher import DES3
from Crypto.Util.Padding import pad

key = b"123456789012345678901234"  # 24-byte key
iv = b"12345678"                   # 8-byte IV

text = input("Enter plaintext: ").encode()

try:
    cipher = DES3.new(key, DES3.MODE_CBC, iv)
    ciphertext = cipher.encrypt(pad(text, 8))

    print("Ciphertext (hex):", ciphertext.hex())

except ValueError as e:
    print("Key error:", e)
```

20. ECB and CBC modes — Error propagation

```python
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad

key = b"0123456789abcdef"
iv = b"abcdef0123456789"
text = b"BlockOneBlockTwoBlockThree"

# ECB encryption
ecb = AES.new(key, AES.MODE_ECB)
c1 = ecb.encrypt(pad(text, 16))

# CBC encryption
cbc = AES.new(key, AES.MODE_CBC, iv)
c2 = cbc.encrypt(pad(text, 16))

# Introduce an error in the first ciphertext block
damaged = bytearray(c2)
damaged[0] ^= 1

# Decrypt original and damaged CBC ciphertext
original = AES.new(key, AES.MODE_CBC, iv).decrypt(c2)
received = AES.new(key, AES.MODE_CBC, iv).decrypt(bytes(damaged))

print("Original plaintext:", original)
print("After ciphertext error:", received)
print("Ciphertext blocks:", len(c2) // 16)
```
